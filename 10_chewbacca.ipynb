# **cgMLST Analysis – chewBBACA**

## Tool Information

- **Tool:** chewBBACA v3.5.1
- **Input:** Genome assemblies (FASTA)
- **Organism:** *Acinetobacter baumannii*
- **Analysis type:** Core genome multilocus sequence typing (cgMLST)

chewBBACA is a gene-by-gene bacterial typing tool used to perform allele-based comparative genomics. It implements the core genome multilocus sequence typing (cgMLST) approach, which compares allelic profiles of conserved genes across multiple bacterial genomes.

Unlike SNP-based phylogenetic analysis, cgMLST focuses on variation within a defined set of core genes shared across isolates. Each gene locus is assigned an allele number, allowing isolates to be compared based on their allelic profiles.

```python
%%bash

# initialise conda
source /home/anaconda/miniconda3/etc/profile.d/conda.sh
conda activate chewie

# verify installation
chewBBACA.py --version
```

## cgMLST Schema Download

Before running cgMLST analysis, a reference schema must be obtained. A schema defines the set of core gene loci used for allele calling across all genomes in the dataset.

In this step, the **cgMLST schema for *Acinetobacter baumannii*** is downloaded using `chewBBACA.py DownloadSchema`.

### What this step does
 
- Creates a directory to store the schema files  
- Downloads the cgMLST schema for *Acinetobacter baumannii*  
- Saves the schema in the specified output directory  

### Key Parameters

- `-sp` → Specifies the organism (*Acinetobacter baumannii*)  
- `-sc` → Specifies the schema type (cgMLST)  
- `-o` → Output directory where the schema will be stored  

The downloaded schema contains the set of reference loci required for allele calling in subsequent chewBBACA analyses.

```python
%%bash
source /home/anaconda/miniconda3/etc/profile.d/conda.sh
conda activate chewie

mkdir -p /data/internship_data/nidhi/aba/new_output/chewie_output/schema
cd /data/internship_data/nidhi/aba/new_output/chewie_output/schema

chewBBACA.py DownloadSchema \
    -sp "Acinetobacter baumannii" \
    -sc cgMLST \
    -o aba_cgmlst
```

## Allele Calling Using cgMLST Schema - Execution

After downloading the cgMLST schema, allele calling is performed for all genome assemblies using `chewBBACA.py AlleleCall`.

This step compares each genome assembly against the reference cgMLST schema to identify alleles present at each locus.

### Key Parameters

- `-i` → Input directory containing genome assemblies  
- `-g` → Directory containing the cgMLST schema  
- `-o` → Output directory for allele calling results  
- `--cpu` → Number of CPU threads used for parallel processing  

The output includes allele profiles for each isolate, which represent the allelic variation across all cgMLST loci and can be used for downstream comparative and population structure analysis.

```python
%%bash
source /home/anaconda/miniconda3/etc/profile.d/conda.sh
conda activate chewie

ASSEMBLY_DIR="/data/internship_data/nidhi/aba/new_output/nextflow_output/assemblies"
SCHEMA_DIR="/data/internship_data/nidhi/aba/new_output/chewie_output/schema/Acinetobacter_baumannii_cgMLST"
OUT_DIR="/data/internship_data/nidhi/aba/new_output/chewie_output"

mkdir -p ${OUT_DIR}

chewBBACA.py AlleleCall \
    -i ${ASSEMBLY_DIR} \
    -g ${SCHEMA_DIR} \
    -o ${OUT_DIR} \
    --cpu 16
```

## Output Files – Allele Calling Results

The `AlleleCall` step generates several result files summarizing allele assignments and quality checks for the analysed genomes.

### Key Output Files

- `results_alleles.tsv` → Allelic profiles for each isolate across all cgMLST loci  
- `results_statistics.tsv` → Summary statistics for allele calling results  
- `results_contigsInfo.tsv` → Contig-level information for each genome  
- `loci_summary_stats.tsv` → Summary statistics for each locus in the schema  
- `cds_coordinates.tsv` → Coordinates of coding sequences identified during allele calling  

### Quality Control Files

- `invalid_cds.txt` → CDS regions that failed validation checks  
- `paralogous_loci.tsv` → Loci identified as paralogous in the dataset  
- `paralogous_counts.tsv` → Counts of paralogous loci per genome  
- `logging_info.txt` → Execution log and runtime information
